# Unicode Card Display Guide

This notebook demonstrates the enhanced Unicode card display capabilities in the Euchre game.

```
# Setup and imports
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from euchre.models import Card, Suit, Rank

print("✅ Unicode display setup complete!")
```

## Card Display Methods

The Card class now provides multiple display methods.

```
# Test different display methods
ace_hearts = Card(Rank.ACE, Suit.HEARTS)
print(f"Standard: {ace_hearts}")
print(f"Unicode: {ace_hearts.unicode_str()}")
print(f"Compact: {ace_hearts.compact_str()}")
```

## Unicode Suit Symbols

Beautiful suit symbols with emoji modifiers.

```
# Test suit symbols
for suit in [Suit.HEARTS, Suit.DIAMONDS, Suit.CLUBS, Suit.SPADES]:
    emoji = Card.get_suit_symbol(suit, True)
    plain = Card.get_suit_symbol(suit, False)
    print(f"{suit.name.title()}: {emoji} / {plain}")
```

## Hand Formatting

Professional hand display with multiple options.

```
# Test hand formatting
hand = [Card(Rank.ACE, Suit.HEARTS), Card(Rank.KING, Suit.SPADES)]
print(f"Hand: {Card.format_hand(hand, use_emoji=True)}")
print(f"Compact: {Card.format_hand(hand, use_emoji=False)}")
```
