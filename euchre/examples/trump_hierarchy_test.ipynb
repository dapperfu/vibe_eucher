# Trump Hierarchy Test

This notebook tests the critical trump hierarchy logic in the Euchre game.

```
# Setup and imports
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from euchre.models import Card, Suit, Rank

print("✅ Trump hierarchy test setup complete!")
```

## Test 1: Basic Trump Hierarchy

Test that Left Bower beats Ace of trump.

```
# Create test cards
right_bower = Card(Rank.JACK, Suit.HEARTS)
left_bower = Card(Rank.JACK, Suit.DIAMONDS)
ace_hearts = Card(Rank.ACE, Suit.HEARTS)

print(f"Right Bower: {right_bower}")
print(f"Left Bower: {left_bower}")
print(f"Ace of Hearts: {ace_hearts}")
```

## Test 2: Trump Values

Check the trump values assigned to each card.

```
# Test trump values
trump_suit = Suit.HEARTS
print(f"Trump suit: {trump_suit.name}")
print(f"Right Bower trump value: {right_bower.get_trump_value(trump_suit)}")
print(f"Left Bower trump value: {left_bower.get_trump_value(trump_suit)}")
print(f"Ace trump value: {ace_hearts.get_trump_value(trump_suit)}")
```

## Test 3: Card Comparison

Test that the comparison logic works correctly.

```
# Test card comparison
left_beats_ace = left_bower.beats(ace_hearts, trump_suit, None)
print(f"Left Bower beats Ace of Hearts: {left_beats_ace} (should be True)")
```
