## AI Factory Pattern

The AI Factory provides a clean way to create different types of AI players.

```
# Create AI factory
ai_factory = AIFactory()
print(f"Available profiles: {ai_factory.get_available_profiles()}")
```

## Create AI Players

Create different AI players with various strategies.

```
# Create AI players
aggressive_alice = ai_factory.create_ai_player("Aggressive_Alice", "aggressive")
conservative_bob = ai_factory.create_ai_player("Conservative_Bob", "conservative")

print(f"Created: {aggressive_alice.name} ({aggressive_alice.ai_profile})")
print(f"Created: {conservative_bob.name} ({conservative_bob.ai_profile})")
```

Analyze how different AI strategies approach the game.
