# Bell Curve Analysis

This notebook analyzes whether game metrics follow normal distributions (bell curves) and identifies outliers.

```python
import sys
from pathlib import Path
import json
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

project_root = Path().resolve().parent
sys.path.insert(0, str(project_root))

# Load data
stats_dir = project_root / "stats"
analysis_files = list(stats_dir.glob("analysis_*.json")) if stats_dir.exists() else []

if analysis_files:
    latest_file = max(analysis_files, key=lambda p: p.stat().st_mtime)
    with open(latest_file, 'r') as f:
        data = json.load(f)
    print(f"Loaded: {latest_file.name}")
else:
    print("No data files found")
    data = None
```

## Normality Tests

```python
if data:
    bell_curve = data.get('bell_curve_analysis', {})
    
    print("=" * 60)
    print("BELL CURVE ANALYSIS")
    print("=" * 60)
    
    for metric_name, analysis in bell_curve.items():
        if 'error' in analysis:
            continue
            
        print(f"\n{analysis.get('name', metric_name).upper()}:")
        print(f"  Mean: {analysis.get('mean', 'N/A'):.2f}")
        print(f"  Std: {analysis.get('std', 'N/A'):.2f}")
        print(f"  Normal Distribution: {analysis.get('is_normal', 'N/A')}")
        print(f"  Outliers (3σ): {analysis.get('outliers_3sigma', 'N/A')} ({analysis.get('outlier_percentage', 0):.2f}%)")
        
        if analysis.get('kolmogorov_smirnov_p_value'):
            print(f"  KS Test p-value: {analysis.get('kolmogorov_smirnov_p_value', 'N/A'):.4f}")
        
        if analysis.get('shapiro_wilk_p_value'):
            print(f"  Shapiro-Wilk p-value: {analysis.get('shapiro_wilk_p_value', 'N/A'):.4f}")
```
