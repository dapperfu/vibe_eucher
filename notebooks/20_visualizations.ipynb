# Visualizations

This notebook creates comprehensive visualizations of the Monte Carlo analysis results.

```python
import sys
from pathlib import Path
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

project_root = Path().resolve().parent
sys.path.insert(0, str(project_root))

# Set style
plt.style.use("seaborn-v0_8")
sns.set_palette("husl")

# Load data
stats_dir = project_root / "stats"
analysis_files = list(stats_dir.glob("analysis_*.json")) if stats_dir.exists() else []

if analysis_files:
    latest_file = max(analysis_files, key=lambda p: p.stat().st_mtime)
    with open(latest_file, "r") as f:
        data = json.load(f)
    print(f"Loaded: {latest_file.name}")
else:
    print("No data files found")
    data = None
```
