# AI Assistant Setup

This notebook sets up the AI Assistant and benchmarks simulation performance.

The AI Assistant uses Monte Carlo simulation through the PyTorch AI to analyze decisions and provide win probability estimates.


```python
import sys
from pathlib import Path

# Add project root to path
project_root = Path().resolve().parent
sys.path.insert(0, str(project_root))

from eucher.ai_players.benchmark_manager import BenchmarkManager
from eucher.ai_players.ai_assistant import AIAssistant
from IPython.display import display, HTML, clear_output
import ipywidgets as widgets
```

```python
# Initialize benchmark manager
benchmark_manager = BenchmarkManager()

# Check if benchmarks exist
benchmarks = benchmark_manager.load_benchmarks()

if benchmarks:
    display(HTML(f"""
    <div style="border: 2px solid #333; padding: 10px; margin: 10px;">
        <h3>Existing Benchmarks</h3>
        <p><strong>Device:</strong> {benchmarks.get('device', 'Unknown')}</p>
        <p><strong>Simulations/Second:</strong> {benchmarks.get('simulations_per_second', 0):.2f}</p>
        <p><strong>Last Updated:</strong> {benchmarks.get('timestamp', 'Unknown')}</p>
    </div>
    """))
else:
    display(HTML("<p>No benchmarks found. Run benchmark below.</p>"))
```

```python
# Run benchmark
benchmark_button = widgets.Button(description="Run Benchmark", button_style='info')
benchmark_output = widgets.Output()

def run_benchmark(b):
    with benchmark_output:
        clear_output()
        display(HTML("<p>Running benchmark... This may take a minute.</p>"))
        
        result = benchmark_manager.benchmark_and_save(num_simulations=1000)
        
        clear_output()
        display(HTML(f"""
        <div style="border: 2px solid #333; padding: 10px; margin: 10px;">
            <h3>Benchmark Results</h3>
            <p><strong>Device:</strong> {result['device']}</p>
            <p><strong>Simulations/Second:</strong> {result['simulations_per_second']:.2f}</p>
            <p><strong>Timestamp:</strong> {result['timestamp']}</p>
        </div>
        """))
        
        # Show thinking time config
        config = benchmark_manager.get_thinking_times_config()
        display(HTML(f"""
        <div style="border: 2px solid #333; padding: 10px; margin: 10px;">
            <h3>Thinking Time Configuration</h3>
            <ul>
                <li><strong>Fast (0.5s):</strong> {config['fast']} simulations</li>
                <li><strong>Quick (1s):</strong> {config['quick']} simulations</li>
                <li><strong>Normal (10s):</strong> {config['normal']} simulations</li>
                <li><strong>Thorough (60s):</strong> {config['thorough']} simulations</li>
                <li><strong>Deep (120s):</strong> {config['deep']} simulations</li>
            </ul>
        </div>
        """))

benchmark_button.on_click(run_benchmark)
display(benchmark_button)
display(benchmark_output)
```
