# Loading Monte Carlo Data

This notebook loads and explores the statistical analysis data from Monte Carlo simulations.

```python
import sys
from pathlib import Path
import json
import pandas as pd
import numpy as np

project_root = Path().resolve().parent
sys.path.insert(0, str(project_root))

print(f"Project root: {project_root}")
```

## Find Analysis Files

```python
# Find all analysis JSON files
stats_dir = project_root / "stats"
analysis_files = list(stats_dir.glob("analysis_*.json")) if stats_dir.exists() else []

print(f"Found {len(analysis_files)} analysis files")
if analysis_files:
    print("\nAvailable files:")
    for f in sorted(analysis_files)[-5:]:  # Show last 5
        print(f"  {f.name}")
    
    # Load most recent
    latest_file = max(analysis_files, key=lambda p: p.stat().st_mtime)
    print(f"\nLoading most recent: {latest_file.name}")
else:
    print("\nNo analysis files found. Run statistical_analysis.py first.")
    latest_file = None
```

## Load Analysis Data

```python
if latest_file:
    with open(latest_file, 'r') as f:
        analysis_data = json.load(f)
    
    print("Data loaded successfully!")
    print(f"\nKeys in analysis data: {list(analysis_data.keys())}")
    
    # Display summary
    summary = analysis_data.get('summary', {})
    print(f"\nSummary:")
    print(f"  Total games: {summary.get('total_games', 'N/A')}")
    print(f"  Team 0 wins: {summary.get('team0_wins', 'N/A')}")
    print(f"  Team 1 wins: {summary.get('team1_wins', 'N/A')}")
else:
    analysis_data = None
    print("No data to load.")
```
