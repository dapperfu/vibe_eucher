# Interactive Full Game

This notebook allows you to play a full game interactively in Jupyter.

You can:
- Play as one player in a 4-player game
- Make all decisions (order up, call trump, play cards)
- See game progress and final results


```python
import sys
from pathlib import Path

# Add project root to path
project_root = Path().resolve().parent
sys.path.insert(0, str(project_root))

from eucher.cards import Card, Deck, Suit
from eucher.game import Game
from eucher.rules import RulesEngine
from eucher.players.base import PlayerProfile
from IPython.display import display, HTML, clear_output
import ipywidgets as widgets
from typing import List, Optional
```

```python
class InteractivePlayerProfile(PlayerProfile):
    """Player profile for interactive gameplay."""
    
    def __init__(self, player_id: int):
        super().__init__()
        self.player_id = player_id
        self.decisions = []
        
    def decide_order_up(self, player, turned_card, dealer_id, trump_suit):
        """Interactive order up decision."""
        # This will be handled by widgets
        return False
    
    def decide_call_trump(self, player, turned_card, trump_suit, must_choose=False):
        """Interactive trump call decision."""
        return None
    
    def choose_card_to_discard(self, player, turned_card=None, ordered_up_by=None):
        """Interactive discard decision."""
        return player.hand[0]  # Default
    
    def play_card(self, player, led_suit, trump_suit, trick_cards, trick_player_ids):
        """Interactive card play."""
        return player.hand[0]  # Default

# Create game with interactive player
player_config = [
    ("You", "simple"),  # You will control this
    ("AI1", "simple"),
    ("AI2", "simple"),
    ("AI3", "simple"),
]

game = Game(player_config)
display(HTML("<h3>Game Started! You are Player 0.</h3>"))
```

## Game Controls

```python
# Game state display
def display_game_state():
    """Display current game state."""
    html = f"""
    <div style="border: 2px solid #333; padding: 10px; margin: 10px;">
        <h3>Game State</h3>
        <p><strong>Scores:</strong> Team 0 = {game.scores[0]}, Team 1 = {game.scores[1]}</p>
        <p><strong>Current Hand:</strong> {len([h for h in game.hands_played if h]) if hasattr(game, 'hands_played') else 'N/A'}</p>
        <p><strong>Dealer:</strong> Player {game.dealer_id}</p>
    </div>
    """
    display(HTML(html))

display_game_state()

# Play next hand button
play_hand_button = widgets.Button(description="Play Next Hand", button_style='success')
hand_output = widgets.Output()

def on_play_hand(b):
    with hand_output:
        clear_output()
        display(HTML("<p>Playing hand...</p>"))
        continue_game = game.play_hand()
        clear_output()
        display_game_state()
        if continue_game:
            winner = game.get_winner()
            if winner is not None:
                display(HTML(f"<h2>Game Over! Team {winner} wins!</h2>"))
            else:
                display(HTML("<p>Hand complete. Click 'Play Next Hand' to continue.</p>"))
        else:
            display(HTML("<p>Game ended.</p>"))

play_hand_button.on_click(on_play_hand)
display(play_hand_button)
display(hand_output)
```
