# EucherGo vs euchre_zero Comparison Analysis

This notebook loads and analyzes the results from Monte Carlo simulations comparing EucherGo and euchre_zero players.


```python
import sys
from pathlib import Path
import json
import pandas as pd
import numpy as np
from typing import Dict, List, Optional

project_root = Path().resolve().parent
sys.path.insert(0, str(project_root))

print(f"Project root: {project_root}")
```

## Find Comparison Results Files


```python
# Find all comparison JSON files
results_dir = project_root / "stats"
comparison_files = list(results_dir.glob("euchergo_vs_euchre_zero_*.json")) if results_dir.exists() else []

print(f"Found {len(comparison_files)} comparison files")
if comparison_files:
    print("\nAvailable files:")
    for f in sorted(comparison_files)[-5:]:  # Show last 5
        print(f"  {f.name}")

    # Load most recent
    latest_file = max(comparison_files, key=lambda p: p.stat().st_mtime)
    print(f"\nLoading most recent: {latest_file.name}")
else:
    print("\nNo comparison files found. Run euchergo_vs_euchre_zero.py first.")
    latest_file = None
```

## Load Comparison Data


```python
if latest_file:
    with open(latest_file, "r") as f:
        comparison_data = json.load(f)

    print("Data loaded successfully!")
    print(f"\nKeys in comparison data: {list(comparison_data.keys())}")

    # Display summary
    print(f"\nSummary:")
    print(f"  Total games: {comparison_data.get('total_games', 'N/A')}")
    print(f"  EucherGo win rate: {comparison_data.get('euchergo', {}).get('win_rate', 0):.1%}")
    print(f"  euchre_zero win rate: {comparison_data.get('euchre_zero', {}).get('win_rate', 0):.1%}")
    print(f"  Average hands per game: {comparison_data.get('avg_hands_per_game', 0):.2f}")
else:
    comparison_data = None
    print("No data to load.")
```

## Create DataFrame for Analysis


```python
if comparison_data:
    # Create DataFrame from game results
    games = comparison_data.get("detailed_statistics", {}).get("games", [])
    
    if games:
        df = pd.DataFrame(games)
        print(f"Loaded {len(df)} games")
        print(f"\nDataFrame columns: {list(df.columns)}")
        print(f"\nFirst few rows:")
        print(df.head())
    else:
        df = None
        print("No game data found in comparison results.")
else:
    df = None
    print("No comparison data available.")
```

## Basic Statistics


```python
if df is not None:
    euchergo_stats = comparison_data.get("euchergo", {})
    euchre_zero_stats = comparison_data.get("euchre_zero", {})
    
    print("=" * 80)
    print("EucherGo Team Statistics")
    print("=" * 80)
    print(f"Games Won: {euchergo_stats.get('games_won', 0)} ({euchergo_stats.get('win_rate', 0):.1%})")
    print(f"Games Lost: {euchergo_stats.get('games_lost', 0)}")
    print(f"Total Score: {euchergo_stats.get('total_score', 0)}")
    print(f"Average Score per Game: {euchergo_stats.get('avg_score', 0):.2f}")
    print(f"Total Tricks Won: {euchergo_stats.get('total_tricks_won', 0)}")
    print(f"Average Tricks per Game: {euchergo_stats.get('avg_tricks_per_game', 0):.2f}")
    print(f"Trump Makes: {euchergo_stats.get('trump_makes', 0)}")
    print(f"Going Alone Success Rate: {euchergo_stats.get('going_alone_success_rate', 0):.1%}")
    
    print("\n" + "=" * 80)
    print("euchre_zero Team Statistics")
    print("=" * 80)
    print(f"Games Won: {euchre_zero_stats.get('games_won', 0)} ({euchre_zero_stats.get('win_rate', 0):.1%})")
    print(f"Games Lost: {euchre_zero_stats.get('games_lost', 0)}")
    print(f"Total Score: {euchre_zero_stats.get('total_score', 0)}")
    print(f"Average Score per Game: {euchre_zero_stats.get('avg_score', 0):.2f}")
    print(f"Total Tricks Won: {euchre_zero_stats.get('total_tricks_won', 0)}")
    print(f"Average Tricks per Game: {euchre_zero_stats.get('avg_tricks_per_game', 0):.2f}")
    print(f"Trump Makes: {euchre_zero_stats.get('trump_makes', 0)}")
    print(f"Going Alone Success Rate: {euchre_zero_stats.get('going_alone_success_rate', 0):.1%}")
    
    print("\n" + "=" * 80)
    print("Comparison")
    print("=" * 80)
    win_rate_diff = euchergo_stats.get('win_rate', 0) - euchre_zero_stats.get('win_rate', 0)
    print(f"Win Rate Difference: {win_rate_diff:+.1%} (EucherGo advantage)")
    avg_score_diff = euchergo_stats.get('avg_score', 0) - euchre_zero_stats.get('avg_score', 0)
    print(f"Average Score Difference: {avg_score_diff:+.2f} (EucherGo advantage)")
    avg_tricks_diff = euchergo_stats.get('avg_tricks_per_game', 0) - euchre_zero_stats.get('avg_tricks_per_game', 0)
    print(f"Average Tricks per Game Difference: {avg_tricks_diff:+.2f} (EucherGo advantage)")
else:
    print("No data available for statistics.")
```

## Per-Player Statistics


```python
if comparison_data:
    player_stats = comparison_data.get("player_stats", {})
    
    print("EucherGo Player Statistics:")
    print("-" * 80)
    euchergo_players = player_stats.get("euchergo", {})
    for player_id, stats in sorted(euchergo_players.items()):
        print(f"\nPlayer {player_id}:")
        print(f"  Tricks Won: {stats.get('tricks_won', 0)}")
        print(f"  Trump Makes: {stats.get('trump_makes', 0)}")
        print(f"  Points Contributed: {stats.get('points_contributed', 0)}")
        print(f"  Going Alone: {stats.get('going_alone_attempts', 0)} attempts, "
              f"{stats.get('going_alone_successes', 0)} successes")
    
    print("\n" + "=" * 80)
    print("euchre_zero Player Statistics:")
    print("-" * 80)
    euchre_zero_players = player_stats.get("euchre_zero", {})
    for player_id, stats in sorted(euchre_zero_players.items()):
        print(f"\nPlayer {player_id}:")
        print(f"  Tricks Won: {stats.get('tricks_won', 0)}")
        print(f"  Trump Makes: {stats.get('trump_makes', 0)}")
        print(f"  Points Contributed: {stats.get('points_contributed', 0)}")
        print(f"  Going Alone: {stats.get('going_alone_attempts', 0)} attempts, "
              f"{stats.get('going_alone_successes', 0)} successes")
else:
    print("No player statistics available.")
```

## Game-by-Game Analysis


```python
if df is not None:
    # Analyze score distributions
    print("Score Distribution:")
    print(f"EucherGo scores: mean={df['euchergo_score'].mean():.2f}, "
          f"std={df['euchergo_score'].std():.2f}, "
          f"min={df['euchergo_score'].min()}, max={df['euchergo_score'].max()}")
    print(f"euchre_zero scores: mean={df['euchre_zero_score'].mean():.2f}, "
          f"std={df['euchre_zero_score'].std():.2f}, "
          f"min={df['euchre_zero_score'].min()}, max={df['euchre_zero_score'].max()}")
    
    print("\nTricks Distribution:")
    print(f"EucherGo tricks: mean={df['euchergo_tricks'].mean():.2f}, "
          f"std={df['euchergo_tricks'].std():.2f}")
    print(f"euchre_zero tricks: mean={df['euchre_zero_tricks'].mean():.2f}, "
          f"std={df['euchre_zero_tricks'].std():.2f}")
    
    print("\nHands Played Distribution:")
    print(f"Mean: {df['hands_played'].mean():.2f}, "
          f"std: {df['hands_played'].std():.2f}, "
          f"min: {df['hands_played'].min()}, max: {df['hands_played'].max()}")
    
    # Win rate by team position
    if 'euchergo_team' in df.columns:
        print("\nWin Rate by EucherGo Team Position:")
        for team_pos in sorted(df['euchergo_team'].unique()):
            team_games = df[df['euchergo_team'] == team_pos]
            wins = (team_games['winner'] == team_pos).sum()
            total = len(team_games)
            print(f"  Team {team_pos}: {wins}/{total} ({wins/total:.1%})")
else:
    print("No game data available for analysis.")
```
